## Separabilidad por señal y condición

```python
sep = pd.read_csv(ARTEFACTOS_TABLAS / 'E4_separabilidad.csv')
mejor = (sep.groupby('signal')['delta_max_abs'].max()
            .sort_values(ascending=False).round(3))
mejor.to_frame('delta máximo')
```
